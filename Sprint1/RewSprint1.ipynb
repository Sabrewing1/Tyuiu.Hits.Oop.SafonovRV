### Вариант задания №


Создайте проект на С# реализующее классы товар (product) Продавец selier, Магазин (store) с использованием геттеров и сеттеров а также одификаторов доступа public, private, internal, От каждого класса по одному экземпляру

```
using System;
using System.Collections.Generic;

// 1. Товар (Product)
public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = category;
        this.description = description;
    }

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value; }
    }

    public int Quantity
    {
        get { return quantity; }
        set { quantity = value; }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public string Description
    {
        get { return description; }
        set { description = value; }
    }

    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        quantity += amount;
    }

    public string GetProductInfo()
    {
        return $"Название: {name}, Цена: {price}, Количество: {quantity}, Категория: {category}, Описание: {description}";
    }
}

// 2. Продавец (Seller)
public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;
    private List<Product> products;

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary;
        this.contactInfo = contactInfo;
        this.products = new List<Product>();
    }

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public decimal Salary
    {
        get { return salary; }
        set { salary = value; }
    }

    public string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
    }

    public void SellProduct(Product product, int quantity)
    {
        if (product.Quantity >= quantity)
        {
            product.UpdateQuantity(-quantity);
            Console.WriteLine($"Продано {quantity} количество товара {product.Name}");
        }
        else
        {
            Console.WriteLine($"Недостаточно товара {product.Name} на складе");
        }
    }

    public void ListProducts()
    {
        Console.WriteLine($"Товары продавца {name}:");
        foreach (var product in products)
        {
            Console.WriteLine(product.GetProductInfo());
        }
    }
}

// 3. Магазин (Store)
public class Store
{
    private string storeName;
    private string location;
    public string storeHours;
    private List<Product> products;
    private List<Seller> sellers;

    public Store(string storeName, string location, string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;
        this.products = new List<Product>();
        this.sellers = new List<Seller>();
    }

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public string StoreHours
    {
        get { return storeHours; }
        set { storeHours = value; }
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
        Console.WriteLine($"Продавец {seller.Name} добавлен в магазин {storeName}");
    }

    public void ListProducts()
    {
        Console.WriteLine($"Товары в магазине {storeName}:");
        foreach (var product in products)
        {
            Console.WriteLine(product.GetProductInfo());
        }
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {storeName}, Адрес: {location}, Часы работы: {storeHours}, Товаров: {products.Count}, Продавцов: {sellers.Count}";
    }
}


{
    
    {
        // Создание экземпляров классов
        Product product1 = new Product("Пылесос", 500m, 10, "Электроника", "Домашний пылесос для трудно доступных мест");
        Seller seller1 = new Seller("Иван Васильевич", "EMP001", 40000m, "+7-999-123-45-67");
        Store store1 = new Store("ТехноСила", "ул. Николая Чаплина, 15", "9:00-21:00");

        // Работа с продуктом
        Console.WriteLine("***** Информация о товаре *****");
        Console.WriteLine(product1.GetProductInfo());
        Console.WriteLine($"Общая стоимость: {product1.GetTotalPrice()}");
        product1.UpdateQuantity(5);
        Console.WriteLine($"После обновления количества: {product1.Quantity}");

        // Работа с продавцом
        Console.WriteLine("\n***** Работа продавца *****");
        seller1.AddProduct(product1);
        seller1.SellProduct(product1, 3);
        seller1.ListProducts();

        // Работа с магазином
        Console.WriteLine("\n***** Работа магазина *****");
        store1.AddProduct(product1);
        store1.AddSeller(seller1);
        store1.ListProducts();
        Console.WriteLine(store1.GetStoreInfo());

        Console.ReadLine();
    }
}
```
