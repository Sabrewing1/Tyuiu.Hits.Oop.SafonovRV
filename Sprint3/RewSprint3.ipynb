Тема: Полиморфизм и Generic классы
Цель задания : Создать программу демонстрирующую принципы полиморфизма перегрузки и перекрытия методов в контексте класса Компьютер и его компонентов Используйте типы Generic для создания гибкой системы.

```
using System;
using System.Collections.Generic;
using System.Linq;

// Базовый абстрактный класс Component
public abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }
    public string Manufacturer { get; set; }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Компонент: {Name}");
        Console.WriteLine($"Производитель: {Manufacturer}");
        Console.WriteLine($"Цена: {Price} руб.");
    }

    public override string ToString()
    {
        return $"{Manufacturer} {Name} - {Price} руб.";
    }
}

// Интерфейс Installable
public interface Installable
{
    void Install();
}

// Класс CPU
public class CPU : Component, Installable
{
    public int Cores { get; set; }
    public double Frequency { get; set; }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Ядра: {Cores}");
        Console.WriteLine($"Частота: {Frequency} ГГц");
    }

    public void Install()
    {
        Console.WriteLine($"Установка процессора {Name} в сокет");
    }

    public static bool operator ==(CPU a, CPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Cores == b.Cores;
    }

    public static bool operator !=(CPU a, CPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is CPU cpu && this == cpu;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Cores);
    }
}

// Класс RAM
public class RAM : Component, Installable
{
    public int Size { get; set; }
    public string Type { get; set; }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Объем: {Size} ГБ");
        Console.WriteLine($"Тип: {Type}");
    }

    public void Install()
    {
        Console.WriteLine($"Установка памяти {Name} в слот DIMM");
    }

    public static bool operator ==(RAM a, RAM b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Size == b.Size;
    }

    public static bool operator !=(RAM a, RAM b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is RAM ram && this == ram;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Size);
    }
}

// Класс GPU
public class GPU : Component, Installable
{
    public int Memory { get; set; }
    public string MemoryType { get; set; }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Видеопамять: {Memory} ГБ");
        Console.WriteLine($"Тип памяти: {MemoryType}");
    }

    public void Install()
    {
        Console.WriteLine($"Установка видеокарты {Name} в слот PCI-E");
    }

    public static bool operator ==(GPU a, GPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Memory == b.Memory;
    }

    public static bool operator !=(GPU a, GPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is GPU gpu && this == gpu;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Memory);
    }
}

// Класс HDD
public class HDD : Component, Installable
{
    public int Capacity { get; set; }
    public int Speed { get; set; }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Емкость: {Capacity} ГБ");
        Console.WriteLine($"Скорость: {Speed} об/мин");
    }

    public void Install()
    {
        Console.WriteLine($"Установка жесткого диска {Name} в отсек");
    }

    public static bool operator ==(HDD a, HDD b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Capacity == b.Capacity;
    }

    public static bool operator !=(HDD a, HDD b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is HDD hdd && this == hdd;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Capacity);
    }
}

// Класс Computer
public class Computer
{
    private List<Component> components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, Installable
    {
        components.Add(component);
        component.Install();
        Console.WriteLine($"Компонент {component.Name} добавлен в компьютер\n");
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        if (components.Remove(component))
        {
            Console.WriteLine($"Компонент {component.Name} удален из компьютера\n");
        }
        else
        {
            Console.WriteLine($"Компонент {component.Name} не найден в компьютере\n");
        }
    }

    public decimal GetTotalPrice()
    {
        return components.Sum(component => component.Price);
    }

    public void ShowComponents()
    {
        Console.WriteLine("Компоненты компьютера:");
        foreach (var component in components)
        {
            Console.WriteLine($" - {component}");
        }
        Console.WriteLine();
    }
}

// Компьютерная фабрика
public static class ComputerFactory
{
    public static Computer CreateGamingPC()
    {
        var computer = new Computer();
        computer.AddComponent(new CPU 
        { 
            Name = "Pentium G4560", 
            Manufacturer = "Intel", 
            Price = 10000, 
            Cores = 2, 
            Frequency = 3.5 
        });
        computer.AddComponent(new GPU 
        { 
            Name = "RTX 3060 Ti", 
            Manufacturer = "NVIDIA", 
            Price = 40000, 
            Memory = 8, 
            MemoryType = "GDDR6" 
        });
        computer.AddComponent(new RAM 
        { 
            Name = "Vengeance", 
            Manufacturer = "Corsair", 
            Price = 10000, 
            Size = 16, 
            Type = "DDR4" 
        });
        return computer;
    }

    public static Computer CreateOfficePC()
    {
        var computer = new Computer();
        computer.AddComponent(new CPU 
        { 
            Name = "Core i5 12400", 
            Manufacturer = "Intel", 
            Price = 20000, 
            Cores = 6, 
            Frequency = 2.5 
        });
        computer.AddComponent(new GPU 
        { 
            Name = "UHD Graphics 730", 
            Manufacturer = "Intel", 
            Price = 5000, 
            Memory = 2, 
            MemoryType = "DDR4" 
        });
        computer.AddComponent(new RAM 
        { 
            Name = "ValueRAM", 
            Manufacturer = "Kingston", 
            Price = 8000, 
            Size = 16, 
            Type = "DDR4" 
        });
        return computer;
    }
}

// Демонстрационное приложение

{
    
    {
        Console.WriteLine("****** Демонстрация работы системы компонентов компьютера ******\n");

        // Создание компьютеров через фабрику
        var gamingPC = ComputerFactory.CreateGamingPC();
        var officePC = ComputerFactory.CreateOfficePC();

        // Демонстрация компонентов
        gamingPC.ShowComponents();
        officePC.ShowComponents();

        // Демонстрация сравнения компонентов
        Console.WriteLine("=== Демонстрация сравнения компонентов ===");
        
        var cpu1 = new CPU { Name = "Pentium G4560", Cores = 2, Manufacturer = "Intel", Price = 10000 };
        var cpu2 = new CPU { Name = "Pentium G4560", Cores = 2, Manufacturer = "Intel", Price = 10000 };
        var cpu3 = new CPU { Name = "Core i5", Cores = 6, Manufacturer = "Intel", Price = 20000 };
        
        Console.WriteLine($"Сравнение одинаковых процессоров: {cpu1 == cpu2}");
        Console.WriteLine($"Сравнение разных процессоров: {cpu1 == cpu3}");

        // Тестирование добавления и удаления компонентов
        Console.WriteLine("\n=== Демонстрация добавления/удаления компонентов ===");
        var testComputer = new Computer();
        var hdd = new HDD 
        { 
            Name = "Barracuda", 
            Manufacturer = "Seagate", 
            Price = 5000, 
            Capacity = 2000, 
            Speed = 7200 
        };

        testComputer.AddComponent(hdd);
        testComputer.ShowComponents();
        testComputer.RemoveComponent(hdd);
        testComputer.ShowComponents();

        // Вывод общей стоимости
        Console.WriteLine($"Стоимость игрового ПК: {gamingPC.GetTotalPrice()} руб.");
        Console.WriteLine($"Стоимость офисного ПК: {officePC.GetTotalPrice()} руб.");

        // Демонстрация полиморфизма
        Console.WriteLine("\n=== Демонстрация полиморфизма ===");
        Component[] components = 
        {
            new CPU { Name = "Core i9", Manufacturer = "Intel", Price = 45000, Cores = 8, Frequency = 5.3 },
            new RAM { Name = "Trident Z", Manufacturer = "G.Skill", Price = 12000, Size = 32, Type = "DDR5" },
            new GPU { Name = "RX 7900 XT", Manufacturer = "AMD", Price = 60000, Memory = 20, MemoryType = "GDDR6" }
        };

        foreach (var component in components)
        {
            component.GetInfo();
            Console.WriteLine();
        }
    }
}
        
```
