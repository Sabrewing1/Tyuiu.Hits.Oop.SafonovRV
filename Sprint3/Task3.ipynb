```
using System;
using System.Collections.Generic;
using System.Linq;

// Generic класс для работы с коллекциями подписок
public class SubscriptionManager<T> where T : Subscription
{
    private List<T> _subscriptions;

    public SubscriptionManager()
    {
        _subscriptions = new List<T>();
    }

    // Добавление подписки
    public void AddSubscription(T subscription)
    {
        _subscriptions.Add(subscription);
        Console.WriteLine($"Подписка {subscription.ServiceName} добавлена в менеджер");
    }

    // Удаление подписки по ID
    public bool RemoveSubscription(string subscriptionId)
    {
        var subscription = _subscriptions.FirstOrDefault(s => s.SubscriptionId == subscriptionId);
        if (subscription != null)
        {
            _subscriptions.Remove(subscription);
            Console.WriteLine($"Подписка {subscription.ServiceName} удалена");
            return true;
        }
        return false;
    }

    // Получение подписки по ID
    public T GetSubscription(string subscriptionId)
    {
        return _subscriptions.FirstOrDefault(s => s.SubscriptionId == subscriptionId);
    }

    // Получение всех активных подписок
    public List<T> GetActiveSubscriptions()
    {
        return _subscriptions.Where(s => s.IsActive).ToList();
    }

    // Обновление стоимости подписки
    public void UpdateSubscriptionCost(string subscriptionId, decimal newCost)
    {
        var subscription = GetSubscription(subscriptionId);
        if (subscription != null)
        {
            subscription.Cost = newCost;
            Console.WriteLine($"Стоимость подписки {subscription.ServiceName} обновлена");
        }
    }

    // Вывод всех подписок
    public void DisplayAllSubscriptions()
    {
        Console.WriteLine($"\n=== Все подписки в менеджере ({typeof(T).Name}) ===");
        foreach (var sub in _subscriptions)
        {
            sub.PrintSubscriptionInfo();
        }
    }

    // Generic метод для фильтрации по условию
    public List<T> FilterSubscriptions(Func<T, bool> predicate)
    {
        return _subscriptions.Where(predicate).ToList();
    }

    // Получение общей стоимости всех подписок
    public decimal GetTotalMonthlyCost()
    {
        return _subscriptions.Sum(s => s.CalculateMonthlyCost());
    }
}

// Базовый класс с новыми атрибутами и методами
public class Subscription
{
    // Новые атрибуты
    public string Category { get; set; }
    public int PriorityLevel { get; set; }
    public string BillingCycle { get; set; }
    public int FreeTrialDays { get; set; }
    public DateTime LastPaymentDate { get; set; }
    public int UsageHours { get; set; }

    // Существующие атрибуты
    public string SubscriptionId { get; set; }
    public string ServiceName { get; set; }
    public decimal Cost { get; set; }
    public DateTime StartDate { get; set; }
    public bool IsActive { get; set; }
    public int DurationMonths { get; set; }
    public string PaymentMethod { get; set; }

    public Subscription(string subscriptionId, string serviceName, decimal cost, 
                       DateTime startDate, int durationMonths, string paymentMethod,
                       string category = "General", int priorityLevel = 1, 
                       string billingCycle = "Monthly", int freeTrialDays = 0)
    {
        SubscriptionId = subscriptionId;
        ServiceName = serviceName;
        Cost = cost;
        StartDate = startDate;
        IsActive = true;
        DurationMonths = durationMonths;
        PaymentMethod = paymentMethod;
        Category = category;
        PriorityLevel = priorityLevel;
        BillingCycle = billingCycle;
        FreeTrialDays = freeTrialDays;
        LastPaymentDate = startDate;
        UsageHours = 0;
    }

    // Новые методы
    public virtual void TrackUsage(int hours)
    {
        UsageHours += hours;
        Console.WriteLine($"Использование услуги {ServiceName}: {hours} часов. Всего: {UsageHours} часов");
    }

    public virtual void ProcessPayment()
    {
        LastPaymentDate = DateTime.Now;
        Console.WriteLine($"Платеж для {ServiceName} обработан. Сумма: {CalculateMonthlyCost():C}");
    }

    public virtual bool IsEligibleForDiscount()
    {
        return DurationMonths >= 12 || PriorityLevel >= 3;
    }

    public virtual void ApplyPromoCode(string promoCode, decimal discount)
    {
        Cost -= discount;
        Console.WriteLine($"Промокод {promoCode} применен. Новая стоимость: {Cost:C}");
    }

    // Перегруженные методы (Overloading)
    public void ExtendSubscription() // Без параметров
    {
        ExtendSubscription(1);
    }

    public void ExtendSubscription(int months) // С одним параметром
    {
        ExtendSubscription(months, false);
    }

    public virtual void ExtendSubscription(int months, bool applyDiscount) // С двумя параметрами
    {
        if (applyDiscount && months >= 6)
        {
            Console.WriteLine($"Применена скидка 10% за продление на {months} месяцев!");
            Cost *= 0.9m;
        }
        DurationMonths += months;
        Console.WriteLine($"Подписка {ServiceName} продлена на {months} месяц(ев)");
    }

    // Переопределяемые методы (Overriding)
    public virtual decimal CalculateMonthlyCost()
    {
        return Cost;
    }

    public virtual decimal CalculateTotalCost()
    {
        return Cost * DurationMonths;
    }

    public virtual void CancelSubscription()
    {
        IsActive = false;
        Console.WriteLine($"Подписка {ServiceName} отменена");
    }

    public virtual void ChangePaymentMethod(string newMethod)
    {
        PaymentMethod = newMethod;
        Console.WriteLine($"Способ оплаты изменен на: {newMethod}");
    }

    public DateTime GetEndDate()
    {
        return StartDate.AddMonths(DurationMonths);
    }

    public virtual string GetSubscriptionDetails()
    {
        return $"ID: {SubscriptionId}, Услуга: {ServiceName}, Стоимость: {Cost:C} в месяц, " +
               $"Активна: {(IsActive ? "Да" : "Нет")}, Длительность: {DurationMonths} мес., " +
               $"Начало: {StartDate:dd.MM.yyyy}, Конец: {GetEndDate():dd.MM.yyyy}, " +
               $"Оплата: {PaymentMethod}, Категория: {Category}, Приоритет: {PriorityLevel}, " +
               $"Цикл оплаты: {BillingCycle}, Пробный период: {FreeTrialDays} дней, " +
               $"Часы использования: {UsageHours}";
    }

    // Новый метод для вывода информации
    public virtual void PrintSubscriptionInfo()
    {
        Console.WriteLine(GetSubscriptionDetails());
    }
}

// Простое наследование с новыми атрибутами
public class OnlineServiceSubscription : Subscription
{
    // Новые атрибуты
    public string ServiceType { get; set; }
    public int ApiCallsLimit { get; set; }
    public bool AutoScaling { get; set; }
    public string SecurityLevel { get; set; }

    // Существующие атрибуты
    public int MaxUsers { get; set; }
    public bool HasCloudStorage { get; set; }
    public int StorageGB { get; set; }

    public OnlineServiceSubscription(string subscriptionId, string serviceName, decimal cost, 
                                   DateTime startDate, int durationMonths, string paymentMethod,
                                   int maxUsers, bool hasCloudStorage, int storageGB,
                                   string serviceType = "SaaS", int apiCallsLimit = 1000,
                                   bool autoScaling = false, string securityLevel = "Standard")
        : base(subscriptionId, serviceName, cost, startDate, durationMonths, paymentMethod, 
               "Online Service", 2, "Monthly", 14)
    {
        MaxUsers = maxUsers;
        HasCloudStorage = hasCloudStorage;
        StorageGB = storageGB;
        ServiceType = serviceType;
        ApiCallsLimit = apiCallsLimit;
        AutoScaling = autoScaling;
        SecurityLevel = securityLevel;
    }

    // Переопределение методов (Overriding)
    public override decimal CalculateMonthlyCost()
    {
        decimal additionalCostPerUser = 2.0m;
        decimal storageCost = HasCloudStorage ? StorageGB * 0.5m : 0;
        decimal apiCost = ApiCallsLimit > 1000 ? (ApiCallsLimit - 1000) * 0.01m : 0;
        decimal scalingCost = AutoScaling ? 10.0m : 0;
        
        return Cost + (MaxUsers > 1 ? (MaxUsers - 1) * additionalCostPerUser : 0) + 
               storageCost + apiCost + scalingCost;
    }

    public override void TrackUsage(int hours)
    {
        base.TrackUsage(hours);
        if (hours > 100)
        {
            Console.WriteLine($"Внимание: высокое использование сервиса {ServiceName}!");
        }
    }

    public override bool IsEligibleForDiscount()
    {
        return base.IsEligibleForDiscount() || ServiceType == "Enterprise";
    }

    // Новые методы
    public void ScaleService(bool enableAutoScaling)
    {
        AutoScaling = enableAutoScaling;
        Console.WriteLine($"Автомасштабирование {(enableAutoScaling ? "включено" : "выключено")}");
    }

    public void IncreaseApiLimit(int additionalCalls)
    {
        ApiCallsLimit += additionalCalls;
        Console.WriteLine($"Лимит API вызовов увеличен до {ApiCallsLimit}");
    }

    public void UpgradeSecurity(string newSecurityLevel)
    {
        SecurityLevel = newSecurityLevel;
        Console.WriteLine($"Уровень безопасности повышен до: {newSecurityLevel}");
    }

    public override string GetSubscriptionDetails()
    {
        return base.GetSubscriptionDetails() + 
               $", Макс. пользователей: {MaxUsers}, " +
               $"Облачное хранилище: {(HasCloudStorage ? "Да" : "Нет")}, " +
               $"Объем: {StorageGB} GB, Тип услуги: {ServiceType}, " +
               $"Лимит API: {ApiCallsLimit}, Автомасштабирование: {(AutoScaling ? "Да" : "Нет")}, " +
               $"Уровень безопасности: {SecurityLevel}";
    }
}

// Сложное наследование с дополнительной функциональностью
public class StreamingSubscription : Subscription
{
    // Новые атрибуты
    public string ContentRegion { get; set; }
    public bool FamilyPlan { get; set; }
    public int SkipLimit { get; set; }
    public string DeviceCompatibility { get; set; }

    // Существующие атрибуты
    public int MaxStreams { get; set; }
    public bool OfflineMode { get; set; }
    public string AudioQuality { get; set; }

    public StreamingSubscription(string subscriptionId, string serviceName, decimal cost,
                               DateTime startDate, int durationMonths, string paymentMethod,
                               int maxStreams, bool offlineMode, string audioQuality,
                               string contentRegion = "Global", bool familyPlan = false,
                               int skipLimit = 6, string deviceCompatibility = "All")
        : base(subscriptionId, serviceName, cost, startDate, durationMonths, paymentMethod,
               "Streaming", 3, "Monthly", 30)
    {
        MaxStreams = maxStreams;
        OfflineMode = offlineMode;
        AudioQuality = audioQuality;
        ContentRegion = contentRegion;
        FamilyPlan = familyPlan;
        SkipLimit = skipLimit;
        DeviceCompatibility = deviceCompatibility;
    }

    // Переопределение с расширенной логикой
    public override void ExtendSubscription(int months, bool applyDiscount)
    {
        base.ExtendSubscription(months, applyDiscount);
        
        if (months >= 6)
        {
            Console.WriteLine($"🎉 Супер предложение! Вы получаете 2 месяца бесплатно!");
            DurationMonths += 2;
            if (FamilyPlan)
            {
                Console.WriteLine("🎁 Бонус: дополнительный поток для семьи!");
                MaxStreams += 1;
            }
        }
    }

    public override void ProcessPayment()
    {
        if (FamilyPlan)
        {
            Console.WriteLine("💳 Применена скидка семейного плана!");
            Cost *= 0.8m; // 20% скидка
        }
        base.ProcessPayment();
    }

    // Новые методы
    public void ChangeRegion(string newRegion)
    {
        ContentRegion = newRegion;
        Console.WriteLine($"Регион контента изменен на: {newRegion}");
    }

    public void EnableFamilyPlan()
    {
        FamilyPlan = true;
        MaxStreams += 2;
        Console.WriteLine("Семейный план активирован! Дополнительные потоки добавлены.");
    }

    public void DownloadForOffline()
    {
        if (OfflineMode)
        {
            Console.WriteLine("Контент загружен для оффлайн прослушивания");
        }
        else
        {
            Console.WriteLine("Оффлайн режим не доступен для вашей подписки");
        }
    }

    public override string GetSubscriptionDetails()
    {
        return base.GetSubscriptionDetails() + 
               $", Макс. потоков: {MaxStreams}, " +
               $"Оффлайн режим: {(OfflineMode ? "Да" : "Нет")}, " +
               $"Качество звука: {AudioQuality}, Регион: {ContentRegion}, " +
               $"Семейный план: {(FamilyPlan ? "Да" : "Нет")}, Лимит пропусков: {SkipLimit}, " +
               $"Совместимость устройств: {DeviceCompatibility}";
    }
}

// Класс с множественным наследованием через интерфейсы
public class VideoSubscription : Subscription, IDownloadable, IShareable
{
    // Новые атрибуты
    public string SupportedResolutions { get; set; }
    public bool DolbyAtmos { get; set; }
    public int ConcurrentStreams { get; set; }
    public string ParentalControl { get; set; }

    // Существующие атрибуты
    public string VideoQuality { get; set; }
    public bool SupportsHDR { get; set; }
    public int MaxDevices { get; set; }

    public VideoSubscription(string subscriptionId, string serviceName, decimal cost,
                           DateTime startDate, int durationMonths, string paymentMethod,
                           string videoQuality, bool supportsHDR, int maxDevices,
                           string supportedResolutions = "1080p,4K", bool dolbyAtmos = false,
                           int concurrentStreams = 3, string parentalControl = "None")
        : base(subscriptionId, serviceName, cost, startDate, durationMonths, paymentMethod,
               "Video Streaming", 4, "Monthly", 7)
    {
        VideoQuality = videoQuality;
        SupportsHDR = supportsHDR;
        MaxDevices = maxDevices;
        SupportedResolutions = supportedResolutions;
        DolbyAtmos = dolbyAtmos;
        ConcurrentStreams = concurrentStreams;
        ParentalControl = parentalControl;
    }

    // Реализация интерфейса IDownloadable
    public bool CanDownload => true;
    
    public void DownloadContent(string contentName)
    {
        Console.WriteLine($"Видео '{contentName}' загружено в максимальном качестве: {VideoQuality}");
        if (DolbyAtmos)
        {
            Console.WriteLine("🎵 Аудио Dolby Atmos включено");
        }
    }

    // Перегруженный метод DownloadContent
    public void DownloadContent(string contentName, string quality)
    {
        Console.WriteLine($"Видео '{contentName}' загружено в качестве: {quality}");
    }

    // Реализация интерфейса IShareable
    public int MaxShares => 3;
    
    public void ShareSubscription(string friendEmail)
    {
        Console.WriteLine($"Приглашение отправлено на {friendEmail}");
        Console.WriteLine($"Осталось приглашений: {MaxShares - 1}");
    }

    public override void ApplyPromoCode(string promoCode, decimal discount)
    {
        if (promoCode.Contains("DOLBY"))
        {
            DolbyAtmos = true;
            Console.WriteLine("Аудио Dolby Atmos активировано!");
        }
        else
        {
            base.ApplyPromoCode(promoCode, discount);
        }
    }

    // Новые методы
    public void SetParentalControl(string level)
    {
        ParentalControl = level;
        Console.WriteLine($"Родительский контроль установлен на уровень: {level}");
    }

    public void ChangeVideoQuality(string newQuality)
    {
        VideoQuality = newQuality;
        Console.WriteLine($"Качество видео изменено на: {newQuality}");
    }

    public override string GetSubscriptionDetails()
    {
        return base.GetSubscriptionDetails() + 
               $", Качество видео: {VideoQuality}, " +
               $"HDR: {(SupportsHDR ? "Да" : "Нет")}, " +
               $"Макс. устройств: {MaxDevices}, Поддерживаемые разрешения: {SupportedResolutions}, " +
               $"Dolby Atmos: {(DolbyAtmos ? "Да" : "Нет")}, Одновременные потоки: {ConcurrentStreams}, " +
               $"Родительский контроль: {ParentalControl}";
    }
}

// Многоуровневое наследование
public class PremiumVideoSubscription : VideoSubscription
{
    // Новые атрибуты
    public bool PersonalCurator { get; set; }
    public int EarlyAccessDays { get; set; }
    public string VIPSupport { get; set; }
    public string ExclusiveEvents { get; set; }

    // Существующие атрибуты
    public bool IncludesExclusiveContent { get; set; }
    public bool EarlyAccess { get; set; }
    public string CustomerTier { get; set; }

    public PremiumVideoSubscription(string subscriptionId, string serviceName, decimal cost,
                                  DateTime startDate, int durationMonths, string paymentMethod,
                                  string videoQuality, bool supportsHDR, int maxDevices,
                                  bool includesExclusiveContent, bool earlyAccess, string customerTier,
                                  bool personalCurator = false, int earlyAccessDays = 7,
                                  string vipSupport = "24/7", string exclusiveEvents = "Premieres")
        : base(subscriptionId, serviceName, cost, startDate, durationMonths, paymentMethod,
               videoQuality, supportsHDR, maxDevices)
    {
        IncludesExclusiveContent = includesExclusiveContent;
        EarlyAccess = earlyAccess;
        CustomerTier = customerTier;
        PersonalCurator = personalCurator;
        EarlyAccessDays = earlyAccessDays;
        VIPSupport = vipSupport;
        ExclusiveEvents = exclusiveEvents;
    }

    // Переопределение с расширенной логикой
    public override decimal CalculateMonthlyCost()
    {
        decimal baseCost = base.CalculateMonthlyCost();
        decimal premiumCost = baseCost;
        
        if (IncludesExclusiveContent) premiumCost += 5.0m;
        if (EarlyAccess) premiumCost += 3.0m;
        if (PersonalCurator) premiumCost += 15.0m;
        if (VIPSupport == "24/7") premiumCost += 10.0m;
        
        // Применяем скидку в зависимости от уровня клиента
        if (CustomerTier == "Platinum") premiumCost *= 0.9m;
        else if (CustomerTier == "Gold") premiumCost *= 0.95m;
        
        return premiumCost;
    }

    public override void ExtendSubscription(int months, bool applyDiscount)
    {
        base.ExtendSubscription(months, applyDiscount);
        if (months >= 12)
        {
            PersonalCurator = true;
            Console.WriteLine("🎁 Вы получаете персонального менеджера на весь срок подписки!");
        }
    }

    // Новые эксклюзивные методы
    public void ScheduleCuratorCall()
    {
        if (PersonalCurator)
        {
            Console.WriteLine("Звонок с персональным куратором запланирован!");
        }
        else
        {
            Console.WriteLine("Услуга персонального куратора не доступна");
        }
    }

    public void AccessVIPEvent(string eventName)
    {
        if (ExclusiveEvents.Contains(eventName))
        {
            Console.WriteLine($"Доступ к VIP событию '{eventName}' предоставлен!");
        }
        else
        {
            Console.WriteLine($"Событие '{eventName}' не доступно в вашем пакете");
        }
    }

    public void AccessExclusiveContent()
    {
        if (IncludesExclusiveContent)
        {
            Console.WriteLine("Доступ к эксклюзивному контенту предоставлен!");
        }
        else
        {
            Console.WriteLine("Эксклюзивный контент не доступен для вашей подписки");
        }
    }

    public override string GetSubscriptionDetails()
    {
        return base.GetSubscriptionDetails() + 
               $", Эксклюзивный контент: {(IncludesExclusiveContent ? "Да" : "Нет")}, " +
               $"Ранний доступ: {(EarlyAccess ? "Да" : "Нет")}, " +
               $"Уровень клиента: {CustomerTier}, Персональный куратор: {(PersonalCurator ? "Да" : "Нет")}, " +
               $"Дни раннего доступа: {EarlyAccessDays}, VIP поддержка: {VIPSupport}, " +
               $"Эксклюзивные события: {ExclusiveEvents}";
    }
}

// Интерфейсы для множественного наследования
public interface IDownloadable
{
    void DownloadContent(string contentName);
    bool CanDownload { get; }
}

public interface IShareable
{
    void ShareSubscription(string friendEmail);
    int MaxShares { get; }
}

// Демонстрация использования

{
    
    {
        Console.WriteLine("=== Демонстрация системы подписок с новыми функциями ===\n");

        // Создание подписок с новыми атрибутами
        var onlineSub = new OnlineServiceSubscription("SUB002", "Онлайн офис", 14.99m, 
            DateTime.Now, 12, "Credit Card", 5, true, 100, "SaaS", 5000, true, "High");

        var streamingSub = new StreamingSubscription("SUB003", "Музыкальный стриминг", 7.99m,
            DateTime.Now.AddDays(7), 6, "PayPal", 3, true, "High Quality", "Europe", true, 10, "Mobile,PC");

        var videoSub = new VideoSubscription("SUB004", "Видео сервис", 12.99m,
            DateTime.Now, 3, "Bank Transfer", "4K Ultra HD", true, 5, "1080p,4K,8K", true, 4, "PG-13");

        var premiumVideoSub = new PremiumVideoSubscription("SUB005", "Премиум видео", 19.99m,
            DateTime.Now, 24, "Credit Card", "8K HDR", true, 10, true, true, "Gold", true, 14, "24/7", "Premieres,Backstage");

        // Демонстрация Generic класса
        var subscriptionManager = new SubscriptionManager<Subscription>();
        
        subscriptionManager.AddSubscription(onlineSub);
        subscriptionManager.AddSubscription(streamingSub);
        subscriptionManager.AddSubscription(videoSub);
        subscriptionManager.AddSubscription(premiumVideoSub);

        // Демонстрация полиморфизма
        Console.WriteLine("\n=== Демонстрация полиморфизма ===");
        List<Subscription> allSubscriptions = new List<Subscription> 
        { 
            onlineSub, streamingSub, videoSub, premiumVideoSub 
        };

        foreach (var sub in allSubscriptions)
        {
            Console.WriteLine($"\n--- {sub.ServiceName} ---");
            sub.PrintSubscriptionInfo();
            Console.WriteLine($"Ежемесячная стоимость: {sub.CalculateMonthlyCost():C}");
            Console.WriteLine($"Общая стоимость: {sub.CalculateTotalCost():C}");
            Console.WriteLine($"Подходит для скидки: {sub.IsEligibleForDiscount()}");
            
            // Демонстрация перегруженных методов
            sub.TrackUsage(50);
            sub.ProcessPayment();
        }

        // Демонстрация переопределения методов
        Console.WriteLine("\n=== Демонстрация переопределения методов ===");
        premiumVideoSub.ExtendSubscription(12, true);
        streamingSub.EnableFamilyPlan();
        videoSub.SetParentalControl("PG-13");

        // Демонстрация Generic менеджера
        Console.WriteLine("\n=== Демонстрация Generic менеджера ===");
        subscriptionManager.DisplayAllSubscriptions();
        
        Console.WriteLine($"\nОбщая ежемесячная стоимость: {subscriptionManager.GetTotalMonthlyCost():C}");
        
        var activeSubs = subscriptionManager.GetActiveSubscriptions();
        Console.WriteLine($"Активных подписок: {activeSubs.Count}");

        // Фильтрация с Generic методом
        var expensiveSubs = subscriptionManager.FilterSubscriptions(s => s.CalculateMonthlyCost() > 15);
        Console.WriteLine($"\nДорогие подписки (>15$): {expensiveSubs.Count}");

        // Демонстрация специализированных менеджеров
        var videoManager = new SubscriptionManager<VideoSubscription>();
        videoManager.AddSubscription(videoSub);
        videoManager.AddSubscription(premiumVideoSub);
        videoManager.DisplayAllSubscriptions();
    }
}
```
