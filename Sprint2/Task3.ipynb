```
using System;

// Базовый класс с новыми атрибутами
public class Subscription
{
    public string SubscriptionId { get; set; }
    public string ServiceName { get; set; }
    public decimal Cost { get; set; }
    public DateTime StartDate { get; set; } // Новый атрибут
    public bool IsActive { get; set; } // Новый атрибут
    public int DurationMonths { get; set; } // Новый атрибут
    public string PaymentMethod { get; set; } // Новый атрибут

    public Subscription(string subscriptionId, string serviceName, decimal cost, 
                       DateTime startDate, int durationMonths, string paymentMethod)
    {
        SubscriptionId = subscriptionId;
        ServiceName = serviceName;
        Cost = cost;
        StartDate = startDate;
        IsActive = true;
        DurationMonths = durationMonths;
        PaymentMethod = paymentMethod;
    }

    public virtual decimal CalculateMonthlyCost()
    {
        return Cost;
    }

    public virtual decimal CalculateTotalCost() // Новый метод
    {
        return Cost * DurationMonths;
    }

    public virtual void ExtendSubscription(int months = 1)
    {
        DurationMonths += months;
        Console.WriteLine($"Подписка {ServiceName} продлена на {months} месяц(ев)");
    }

    public virtual void CancelSubscription() // Новый метод
    {
        IsActive = false;
        Console.WriteLine($"Подписка {ServiceName} отменена");
    }

    public virtual void ChangePaymentMethod(string newMethod) // Новый метод
    {
        PaymentMethod = newMethod;
        Console.WriteLine($"Способ оплаты изменен на: {newMethod}");
    }

    public DateTime GetEndDate() // Новый метод
    {
        return StartDate.AddMonths(DurationMonths);
    }

    public virtual string GetSubscriptionDetails()
    {
        return $"ID: {SubscriptionId}, Услуга: {ServiceName}, Стоимость: {Cost:C} в месяц, " +
               $"Активна: {(IsActive ? "Да" : "Нет")}, Длительность: {DurationMonths} мес., " +
               $"Начало: {StartDate:dd.MM.yyyy}, Конец: {GetEndDate():dd.MM.yyyy}, " +
               $"Оплата: {PaymentMethod}";
    }
}

// Простое наследование
public class OnlineServiceSubscription : Subscription
{
    public int MaxUsers { get; set; }
    public bool HasCloudStorage { get; set; } // Новый атрибут
    public int StorageGB { get; set; } // Новый атрибут

    public OnlineServiceSubscription(string subscriptionId, string serviceName, decimal cost, 
                                   DateTime startDate, int durationMonths, string paymentMethod,
                                   int maxUsers, bool hasCloudStorage, int storageGB)
        : base(subscriptionId, serviceName, cost, startDate, durationMonths, paymentMethod)
    {
        MaxUsers = maxUsers;
        HasCloudStorage = hasCloudStorage;
        StorageGB = storageGB;
    }

    public override decimal CalculateMonthlyCost()
    {
        decimal additionalCostPerUser = 2.0m;
        decimal storageCost = HasCloudStorage ? StorageGB * 0.5m : 0;
        return Cost + (MaxUsers > 1 ? (MaxUsers - 1) * additionalCostPerUser : 0) + storageCost;
    }

    public void IncreaseStorage(int additionalGB) // Новый метод
    {
        StorageGB += additionalGB;
        Console.WriteLine($"Объем хранилища увеличен на {additionalGB} GB");
    }

    public override string GetSubscriptionDetails()
    {
        return base.GetSubscriptionDetails() + 
               $", Макс. пользователей: {MaxUsers}, " +
               $"Облачное хранилище: {(HasCloudStorage ? "Да" : "Нет")}, " +
               $"Объем: {StorageGB} GB";
    }
}

// Сложное наследование с дополнительной функциональностью
public class StreamingSubscription : Subscription
{
    public int MaxStreams { get; set; }
    public bool OfflineMode { get; set; } // Новый атрибут
    public string AudioQuality { get; set; } // Новый атрибут

    public StreamingSubscription(string subscriptionId, string serviceName, decimal cost,
                               DateTime startDate, int durationMonths, string paymentMethod,
                               int maxStreams, bool offlineMode, string audioQuality)
        : base(subscriptionId, serviceName, cost, startDate, durationMonths, paymentMethod)
    {
        MaxStreams = maxStreams;
        OfflineMode = offlineMode;
        AudioQuality = audioQuality;
    }

    public override void ExtendSubscription(int months = 1)
    {
        base.ExtendSubscription(months);
        
        if (months >= 6)
        {
            Console.WriteLine($"🎉 Супер предложение! Вы получаете 2 месяца бесплатно!");
            DurationMonths += 2; // Добавляем бонусные месяцы
        }
        else if (months >= 3)
        {
            Console.WriteLine($"🎉 Специальное предложение! Вы получаете скидку 15%!");
            // Применяем скидку к следующему платежу
        }
    }

    public void DownloadForOffline() // Новый метод
    {
        if (OfflineMode)
        {
            Console.WriteLine("Контент загружен для оффлайн прослушивания");
        }
        else
        {
            Console.WriteLine("Оффлайн режим не доступен для вашей подписки");
        }
    }

    public override string GetSubscriptionDetails()
    {
        return base.GetSubscriptionDetails() + 
               $", Макс. потоков: {MaxStreams}, " +
               $"Оффлайн режим: {(OfflineMode ? "Да" : "Нет")}, " +
               $"Качество звука: {AudioQuality}";
    }
}

// Множественное наследование через интерфейсы
public interface IDownloadable
{
    void DownloadContent(string contentName);
    bool CanDownload { get; }
}

public interface IShareable
{
    void ShareSubscription(string friendEmail);
    int MaxShares { get; }
}

// Класс с множественным наследованием (через интерфейсы)
public class VideoSubscription : Subscription, IDownloadable, IShareable
{
    public string VideoQuality { get; set; }
    public bool SupportsHDR { get; set; } // Новый атрибут
    public int MaxDevices { get; set; } // Новый атрибут

    public VideoSubscription(string subscriptionId, string serviceName, decimal cost,
                           DateTime startDate, int durationMonths, string paymentMethod,
                           string videoQuality, bool supportsHDR, int maxDevices)
        : base(subscriptionId, serviceName, cost, startDate, durationMonths, paymentMethod)
    {
        VideoQuality = videoQuality;
        SupportsHDR = supportsHDR;
        MaxDevices = maxDevices;
    }

    // Реализация интерфейса IDownloadable
    public bool CanDownload => true;
    
    public void DownloadContent(string contentName)
    {
        Console.WriteLine($"Видео '{contentName}' загружено в максимальном качестве: {VideoQuality}");
    }

    // Реализация интерфейса IShareable
    public int MaxShares => 3;
    
    public void ShareSubscription(string friendEmail)
    {
        Console.WriteLine($"Приглашение отправлено на {friendEmail}");
        Console.WriteLine($"Осталось приглашений: {MaxShares - 1}");
    }

    public void ChangeVideoQuality(string newQuality) // Новый метод
    {
        VideoQuality = newQuality;
        Console.WriteLine($"Качество видео изменено на: {newQuality}");
    }

    public override string GetSubscriptionDetails()
    {
        return base.GetSubscriptionDetails() + 
               $", Качество видео: {VideoQuality}, " +
               $"HDR: {(SupportsHDR ? "Да" : "Нет")}, " +
               $"Макс. устройств: {MaxDevices}";
    }
}

// Многоуровневое наследование
public class PremiumVideoSubscription : VideoSubscription
{
    public bool IncludesExclusiveContent { get; set; }
    public bool EarlyAccess { get; set; } // Новый атрибут
    public string CustomerTier { get; set; } // Новый атрибут

    public PremiumVideoSubscription(string subscriptionId, string serviceName, decimal cost,
                                  DateTime startDate, int durationMonths, string paymentMethod,
                                  string videoQuality, bool supportsHDR, int maxDevices,
                                  bool includesExclusiveContent, bool earlyAccess, string customerTier)
        : base(subscriptionId, serviceName, cost, startDate, durationMonths, paymentMethod,
               videoQuality, supportsHDR, maxDevices)
    {
        IncludesExclusiveContent = includesExclusiveContent;
        EarlyAccess = earlyAccess;
        CustomerTier = customerTier;
    }

    public override decimal CalculateMonthlyCost()
    {
        decimal baseCost = base.CalculateMonthlyCost();
        decimal premiumCost = baseCost;
        
        if (IncludesExclusiveContent) premiumCost += 5.0m;
        if (EarlyAccess) premiumCost += 3.0m;
        if (CustomerTier == "Platinum") premiumCost += 10.0m;
        else if (CustomerTier == "Gold") premiumCost += 5.0m;
        
        return premiumCost;
    }

    public void AccessExclusiveContent() // Новый метод
    {
        if (IncludesExclusiveContent)
        {
            Console.WriteLine("Доступ к эксклюзивному контенту предоставлен!");
        }
        else
        {
            Console.WriteLine("Эксклюзивный контент не доступен для вашей подписки");
        }
    }

    public override void ExtendSubscription(int months = 1)
    {
        base.ExtendSubscription(months);
        if (months >= 12)
        {
            Console.WriteLine("🎁 Вы получаете персонального менеджера на весь срок подписки!");
        }
    }

    public override string GetSubscriptionDetails()
    {
        return base.GetSubscriptionDetails() + 
               $", Эксклюзивный контент: {(IncludesExclusiveContent ? "Да" : "Нет")}, " +
               $"Ранний доступ: {(EarlyAccess ? "Да" : "Нет")}, " +
               $"Уровень клиента: {CustomerTier}";
    }
}

// Дополнительный класс для демонстрации композиции
public class SubscriptionBundle
{
    private List<Subscription> subscriptions;
    public string BundleName { get; set; }
    public decimal BundleDiscount { get; set; }

    public SubscriptionBundle(string bundleName, decimal discount)
    {
        subscriptions = new List<Subscription>();
        BundleName = bundleName;
        BundleDiscount = discount;
    }

    public void AddSubscription(Subscription subscription)
    {
        subscriptions.Add(subscription);
        Console.WriteLine($"Подписка {subscription.ServiceName} добавлена в пакет {BundleName}");
    }

    public decimal CalculateTotalBundleCost()
    {
        decimal total = subscriptions.Sum(sub => sub.CalculateMonthlyCost());
        return total * (1 - BundleDiscount / 100);
    }

    public void DisplayBundleInfo()
    {
        Console.WriteLine($"\n=== Пакет подписок: {BundleName} ===");
        Console.WriteLine($"Скидка: {BundleDiscount}%");
        Console.WriteLine("Включает подписки:");
        foreach (var sub in subscriptions)
        {
            Console.WriteLine($"- {sub.ServiceName}");
        }
        Console.WriteLine($"Общая стоимость с учетом скидки: {CalculateTotalBundleCost():C}");
    }
}


{
    
    {
        // Создание объектов с новыми атрибутами
        Subscription onlineSub = new OnlineServiceSubscription("SUB002", "Онлайн офис", 14.99m, 
            DateTime.Now, 12, "Credit Card", 5, true, 100);
        
        Subscription streamingSub = new StreamingSubscription("SUB003", "Музыкальный стриминг", 7.99m,
            DateTime.Now.AddDays(7), 6, "PayPal", 3, true, "High Quality");
        
        VideoSubscription videoSub = new VideoSubscription("SUB004", "Видео сервис", 12.99m,
            DateTime.Now, 3, "Bank Transfer", "4K Ultra HD", true, 5);
        
        PremiumVideoSubscription premiumVideoSub = new PremiumVideoSubscription("SUB005", "Премиум видео", 19.99m,
            DateTime.Now, 24, "Credit Card", "8K HDR", true, 10, true, true, "Gold");

        // Демонстрация новых методов
        Console.WriteLine("=== Демонстрация новых возможностей ===");
        
        Console.WriteLine("\n" + onlineSub.GetSubscriptionDetails());
        ((OnlineServiceSubscription)onlineSub).IncreaseStorage(50);
        onlineSub.ExtendSubscription(6);
        Console.WriteLine($"Общая стоимость: {onlineSub.CalculateTotalCost():C}");
        
        Console.WriteLine("\n" + streamingSub.GetSubscriptionDetails());
        ((StreamingSubscription)streamingSub).DownloadForOffline();
        streamingSub.ChangePaymentMethod("Apple Pay");
        
        Console.WriteLine("\n" + videoSub.GetSubscriptionDetails());
        videoSub.DownloadContent("Новый блокбастер");
        videoSub.ShareSubscription("friend@example.com");
        
        Console.WriteLine("\n" + premiumVideoSub.GetSubscriptionDetails());
        premiumVideoSub.AccessExclusiveContent();
        Console.WriteLine($"Стоимость с учетом всех премиум функций: {premiumVideoSub.CalculateMonthlyCost():C}");
        
        // Демонстрация отмены подписки
        Console.WriteLine("\n=== Демонстрация отмены подписки ===");
        Subscription tempSub = new Subscription("TEMP001", "Тестовая услуга", 9.99m, 
            DateTime.Now, 1, "Cash");
        Console.WriteLine(tempSub.GetSubscriptionDetails());
        tempSub.CancelSubscription();
        Console.WriteLine(tempSub.GetSubscriptionDetails());
        
        // Демонстрация пакета подписок
        Console.WriteLine("\n=== Демонстрация пакета подписок ===");
        SubscriptionBundle bundle = new SubscriptionBundle("Медиа пакет", 15);
        bundle.AddSubscription(streamingSub);
        bundle.AddSubscription(videoSub);
        bundle.AddSubscription(premiumVideoSub);
        bundle.DisplayBundleInfo();
    }
}
```
