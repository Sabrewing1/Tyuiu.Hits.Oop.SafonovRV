### Спринт Ревью 2


Тема: Наследование, Абстрактный класс интерфейсы
Цель задания : создать классы, представляющие структуру коммерческого предприятия, используя, сложное и множественное наследование

```
using System;
using System.Collections.Generic;

// Базовый класс Employee
public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public double Salary { get; set; }

    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void Plaintiff()
    {
        Console.WriteLine($"Сотрудник: {Name}");
        Console.WriteLine($"Должность: {Position}");
        Console.WriteLine($"Зарплата: {Salary:C}");
    }
}

// Класс Manager с простым наследованием
public class Manager : Employee
{
    public string Department { get; set; }

    public Manager(string name, string position, double salary, string department) 
        : base(name, position, salary)
    {
        Department = department;
    }

    public override void Plaintiff()
    {
        base.Plaintiff();
        Console.WriteLine($"Отдел: {Department}");
    }
}

// Интерфейс IReportable
public interface IReportable
{
    string GenerateReport();
}

// Интерфейс ITeamManagement
public interface ITeamManagement
{
    string[] GetTeamMembers();
}

// Класс ProjectManager со сложным наследованием и множественной реализацией интерфейсов
public class ProjectManager : Manager, IReportable, ITeamManagement
{
    public string ProjectName { get; set; }
    private List<string> teamMembers;

    public ProjectManager(string name, string position, double salary, string department, string projectName)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
        teamMembers = new List<string>();
    }

    public void AddTeamMember(string member)
    {
        teamMembers.Add(member);
    }

    public override void Plaintiff()
    {
        base.Plaintiff();
        Console.WriteLine($"Проект: {ProjectName}");
    }

    // Реализация интерфейса IReportable
    public string GenerateReport()
    {
        return $"Отчет по проекту {ProjectName}. Статус: в работе. Команда: {teamMembers.Count} человек.";
    }

    // Реализация интерфейса ITeamManagement
    public string[] GetTeamMembers()
    {
        return teamMembers.ToArray();
    }
}


{
    
    {
        // Тестирование классов
        Console.WriteLine("****** Тестирование наследования ******");
        
        // Создание экземпляра Employee
        Employee employee = new Employee("Николай Васильевич", "Разработчик", 50000);
        Console.WriteLine("\n****** Employee ******");
        employee.Plaintiff();

        // Создание экземпляра Manager
        Manager manager = new Manager("Петя Сидоров", "Менеджер", 70000, "IT");
        Console.WriteLine("\n****** Manager ******");
        manager.Plaintiff();

        // Создание экземпляра ProjectManager
        ProjectManager projectManager = new ProjectManager("Сергей Козлов", "Проектный менеджер", 
                                                          90000, "Разработка", "Новый веб-портал");
        
        // Добавляем членов команды
        projectManager.AddTeamMember("Александр Петров");
        projectManager.AddTeamMember("Валентина Рыбкина");
        projectManager.AddTeamMember("Алексей Смирнов");

        Console.WriteLine("\n--- ProjectManager ---");
        projectManager.Plaintiff();

        // Тестирование интерфейсов
        Console.WriteLine("\n****** Тестирование интерфейсов ******");
        
        // Тестирование IReportable
        IReportable reportable = projectManager;
        Console.WriteLine($"Отчет: {reportable.GenerateReport()}");

        // Тестирование ITeamManagement
        ITeamManagement teamManagement = projectManager;
        string[] team = teamManagement.GetTeamMembers();
        Console.WriteLine("Члены команды:");
        foreach (var member in team)
        {
            Console.WriteLine($" - {member}");
        }

        // Демонстрация полиморфизма
        Console.WriteLine("\n****** Полиморфизм ******");
        Employee[] employees = { employee, manager, projectManager };
        
        foreach (var emp in employees)
        {
            Console.WriteLine("\n****** Полиморфный вызов ******");
            emp.Plaintiff();
        }
    }
}
        
```
