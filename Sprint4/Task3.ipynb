```
using System;
using System.Collections.Generic;
using System.Linq;

// Базовый класс подписки
public class Subscription
{
    public string SubscriptionId { get; set; }
    public string ServiceName { get; set; }
    public decimal Cost { get; set; }
    public bool IsActive { get; set; }
    // Новые атрибуты
    public string Language { get; set; }
    public string Theme { get; set; }
    public int Rating { get; set; }

    public Subscription(string subscriptionId, string serviceName, decimal cost)
    {
        SubscriptionId = subscriptionId;
        ServiceName = serviceName;
        Cost = cost;
        IsActive = true;
        Language = "English";
        Theme = "Light";
        Rating = 0;
    }

    // Новые методы
    public virtual void ChangeLanguage(string newLanguage)
    {
        Language = newLanguage;
        Console.WriteLine($"Язык изменен на: {newLanguage}");
    }
    
    public virtual void ChangeTheme(string newTheme)
    {
        Theme = newTheme;
        Console.WriteLine($"Тема изменена на: {newTheme}");
    }
    
    public virtual void RateSubscription(int rating)
    {
        Rating = rating;
        Console.WriteLine($"Подписка оценена на {rating} ⭐");
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"{ServiceName} - {Cost:C} - {Language} - {Theme} - {Rating}⭐");
    }
}

// Generic менеджер подписок
public class SubscriptionManager<T> where T : Subscription
{
    private List<T> _subscriptions = new List<T>();

    public void AddSubscription(T subscription)
    {
        _subscriptions.Add(subscription);
        Console.WriteLine($"Добавлена: {subscription.ServiceName}");
    }

    public void DisplayAll()
    {
        Console.WriteLine($"\n=== Все подписки ({_subscriptions.Count}) ===");
        foreach (var sub in _subscriptions)
        {
            sub.PrintInfo();
        }
    }
}

// Демонстрация
class Program
{
    static void Main(string[] args)
    {
        Console.WriteLine("=== ДЕМОНСТРАЦИЯ СИСТЕМЫ ПОДПИСОК ===\n");

        // Создание менеджера
        var manager = new SubscriptionManager<Subscription>();

        // Создание подписок
        var sub1 = new Subscription("SUB1", "Музыка", 9.99m) 
        { 
            Language = "Russian", 
            Theme = "Dark" 
        };
        
        var sub2 = new Subscription("SUB2", "Видео", 14.99m) 
        { 
            Language = "English", 
            Theme = "Blue" 
        };

        // Демонстрация методов
        sub1.RateSubscription(5);
        sub2.ChangeLanguage("Spanish");
        sub2.ChangeTheme("Dark");

        // Добавление в менеджер
        manager.AddSubscription(sub1);
        manager.AddSubscription(sub2);

        // Показать все
        manager.DisplayAll();

        Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ЗАВЕРШЕНА ===");
        Console.WriteLine("Нажмите любую клавишу для выхода...");
        Console.ReadKey();
    }
}
```
