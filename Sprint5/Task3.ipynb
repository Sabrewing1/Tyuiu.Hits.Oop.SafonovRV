```
using System;
using System.Collections.Generic;
using System.Linq;

// Делегаты для работы с подписками
public delegate void SubscriptionHandler<T>(T subscription);
public delegate void SubscriptionAction(Subscription subscription);

// Базовый класс подписки
public class Subscription
{
    private decimal _cost;
    
    public string SubscriptionId { get; set; }
    public string ServiceName { get; set; }
    public decimal Cost 
    { 
        get => _cost;
        set
        {
            if (value != _cost)
            {
                _cost = value;
                OnCostChanged?.Invoke(this);
            }
        }
    }
    public bool IsActive { get; set; }
    public string Language { get; set; }
    public string Theme { get; set; }
    public int Rating { get; set; }
    public DateTime CreatedDate { get; set; }
    public DateTime ExpiryDate { get; set; }
    public List<string> Features { get; set; }
    public TimeSpan WatchTime { get; set; }

    // События
    public event Action<Subscription> OnCostChanged;
    public event Action<Subscription> OnSubscriptionExpired;

    public Subscription(string subscriptionId, string serviceName, decimal cost)
    {
        SubscriptionId = subscriptionId;
        ServiceName = serviceName;
        Cost = cost;
        IsActive = true;
        Language = "English";
        Theme = "Light";
        Rating = 0;
        CreatedDate = DateTime.Now;
        ExpiryDate = DateTime.Now.AddMonths(1);
        Features = new List<string>();
        WatchTime = TimeSpan.Zero;
    }

    public virtual void ChangeLanguage(string newLanguage)
    {
        Language = newLanguage;
        Console.WriteLine($"Язык изменен на: {newLanguage}");
    }
    
    public virtual void ChangeTheme(string newTheme)
    {
        Theme = newTheme;
        Console.WriteLine($"Тема изменена на: {newTheme}");
    }
    
    public virtual void RateSubscription(int rating)
    {
        if (rating < 0 || rating > 5)
            throw new ArgumentException("Рейтинг должен быть от 0 до 5");
        
        Rating = rating;
        Console.WriteLine($"Подписка оценена на {rating} ⭐");
    }

    // Новые методы
    public virtual void AddFeature(string feature)
    {
        Features.Add(feature);
        Console.WriteLine($"Добавлена функция: {feature}");
    }

    public virtual void AddWatchTime(TimeSpan time)
    {
        WatchTime += time;
        Console.WriteLine($"Добавлено время просмотра: {time}. Всего: {WatchTime}");
    }

    public virtual bool CheckExpiry()
    {
        if (DateTime.Now > ExpiryDate && IsActive)
        {
            IsActive = false;
            OnSubscriptionExpired?.Invoke(this);
            return true;
        }
        return false;
    }

    public virtual void ExtendSubscription(int months)
    {
        ExpiryDate = ExpiryDate.AddMonths(months);
        IsActive = true;
        Console.WriteLine($"Подписка продлена до: {ExpiryDate:d}");
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"{ServiceName} - {Cost:C} - {Language} - {Theme} - {Rating}⭐");
        Console.WriteLine($"   Активна: {IsActive} | Создана: {CreatedDate:d} | Истекает: {ExpiryDate:d}");
        Console.WriteLine($"   Время просмотра: {WatchTime} | Функции: {string.Join(", ", Features)}");
    }
}

// Производные классы подписок
public class StreamingSubscription : Subscription
{
    public int Quality { get; set; } // 720, 1080, 4K
    public bool OfflineMode { get; set; }
    public int SimultaneousStreams { get; set; }
    public List<string> FavoriteGenres { get; set; }

    public StreamingSubscription(string id, string name, decimal cost) 
        : base(id, name, cost)
    {
        Quality = 1080;
        OfflineMode = false;
        SimultaneousStreams = 1;
        FavoriteGenres = new List<string>();
    }

    public void ChangeQuality(int quality)
    {
        int[] allowedQualities = { 720, 1080, 2160 };
        if (allowedQualities.Contains(quality))
        {
            Quality = quality;
            Console.WriteLine($"Качество изменено на: {quality}p");
        }
        else
        {
            Console.WriteLine("Недопустимое качество. Допустимые значения: 720, 1080, 2160");
        }
    }

    public void ToggleOfflineMode()
    {
        OfflineMode = !OfflineMode;
        Console.WriteLine($"Офлайн-режим: {(OfflineMode ? "Включен" : "Выключен")}");
    }

    public void AddFavoriteGenre(string genre)
    {
        FavoriteGenres.Add(genre);
        Console.WriteLine($"Добавлен любимый жанр: {genre}");
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"   Качество: {Quality}p | Офлайн-режим: {OfflineMode}");
        Console.WriteLine($"   Одновременные потоки: {SimultaneousStreams}");
        Console.WriteLine($"   Любимые жанры: {string.Join(", ", FavoriteGenres)}");
    }
}

public class MusicSubscription : Subscription
{
    public bool HighQualityAudio { get; set; }
    public List<string> Playlists { get; set; }
    public List<string> DownloadedAlbums { get; set; }
    public string ListeningMode { get; set; }

    public MusicSubscription(string id, string name, decimal cost) 
        : base(id, name, cost)
    {
        HighQualityAudio = true;
        Playlists = new List<string>();
        DownloadedAlbums = new List<string>();
        ListeningMode = "Normal";
    }

    public void ToggleAudioQuality()
    {
        HighQualityAudio = !HighQualityAudio;
        Console.WriteLine($"Высокое качество аудио: {(HighQualityAudio ? "Включено" : "Выключено")}");
    }

    public void CreatePlaylist(string playlistName)
    {
        Playlists.Add(playlistName);
        Console.WriteLine($"Создан плейлист: {playlistName}");
    }

    public void DownloadAlbum(string albumName)
    {
        DownloadedAlbums.Add(albumName);
        Console.WriteLine($"Альбом загружен: {albumName}");
    }

    public void ChangeListeningMode(string mode)
    {
        ListeningMode = mode;
        Console.WriteLine($"Режим прослушивания изменен на: {mode}");
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"   HQ Аудио: {HighQualityAudio} | Режим: {ListeningMode}");
        Console.WriteLine($"   Плейлисты: {Playlists.Count} | Загружено альбомов: {DownloadedAlbums.Count}");
    }
}

// Generic менеджер подписок с событиями
public class SubscriptionManager<T> where T : Subscription
{
    private Dictionary<string, T> _subscriptions = new Dictionary<string, T>();
    
    // События
    public event SubscriptionHandler<T> OnSubscriptionAdded;
    public event SubscriptionHandler<T> OnSubscriptionRemoved;
    public event Action<string, decimal, decimal> OnCostChanged;

    public void AddSubscription(T subscription)
    {
        _subscriptions[subscription.SubscriptionId] = subscription;
        subscription.OnCostChanged += (sub) => 
            OnCostChanged?.Invoke(sub.ServiceName, sub.Cost, 0);
        
        OnSubscriptionAdded?.Invoke(subscription);
        Console.WriteLine($"Добавлена: {subscription.ServiceName}");
    }

    public bool RemoveSubscription(string subscriptionId)
    {
        if (_subscriptions.TryGetValue(subscriptionId, out T subscription))
        {
            _subscriptions.Remove(subscriptionId);
            OnSubscriptionRemoved?.Invoke(subscription);
            Console.WriteLine($"Удалена: {subscription.ServiceName}");
            return true;
        }
        return false;
    }

    public T GetSubscription(string subscriptionId)
    {
        return _subscriptions.GetValueOrDefault(subscriptionId);
    }

    public IEnumerable<T> GetActiveSubscriptions()
    {
        return _subscriptions.Values.Where(sub => sub.IsActive);
    }

    public IEnumerable<T> GetExpiredSubscriptions()
    {
        return _subscriptions.Values.Where(sub => !sub.IsActive);
    }

    public decimal GetTotalMonthlyCost()
    {
        return _subscriptions.Values.Where(sub => sub.IsActive).Sum(sub => sub.Cost);
    }

    public void DisplayAll()
    {
        Console.WriteLine($"\n=== Все подписки ({_subscriptions.Count}) ===");
        foreach (var sub in _subscriptions.Values)
        {
            sub.PrintInfo();
            Console.WriteLine();
        }
    }

    public void CheckAllExpirations()
    {
        Console.WriteLine("\n=== Проверка истечения срока подписок ===");
        foreach (var sub in _subscriptions.Values)
        {
            sub.CheckExpiry();
        }
    }
}

// Сервис уведомлений
public class NotificationService
{
    public void SendSubscriptionAddedNotification(Subscription subscription)
    {
        Console.WriteLine($"🔔 УВЕДОМЛЕНИЕ: Добавлена новая подписка '{subscription.ServiceName}'");
    }

    public void SendSubscriptionRemovedNotification(Subscription subscription)
    {
        Console.WriteLine($"🔔 УВЕДОМЛЕНИЕ: Подписка '{subscription.ServiceName}' удалена");
    }

    public void SendCostChangedNotification(string serviceName, decimal oldCost, decimal newCost)
    {
        Console.WriteLine($"🔔 УВЕДОМЛЕНИЕ: Цена '{serviceName}' изменена с {oldCost:C} на {newCost:C}");
    }

    public void SendExpirationNotification(Subscription subscription)
    {
        Console.WriteLine($"🔔 УВЕДОМЛЕНИЕ: Подписка '{subscription.ServiceName}' истекла!");
    }
}

// Демонстрация

{

    {
        Console.WriteLine("=== РАСШИРЕННАЯ СИСТЕМА ПОДПИСОК ===\n");

        // Создание менеджера
        var manager = new SubscriptionManager<Subscription>();
        var notifications = new NotificationService();

        // Подписка на события
        manager.OnSubscriptionAdded += notifications.SendSubscriptionAddedNotification;
        manager.OnSubscriptionRemoved += notifications.SendSubscriptionRemovedNotification;
        manager.OnCostChanged += notifications.SendCostChangedNotification;

        // Создание различных подписок
        var streaming = new StreamingSubscription("SUB1", "Premium Video", 15.99m);
        streaming.OnSubscriptionExpired += notifications.SendExpirationNotification;
        
        var music = new MusicSubscription("SUB2", "Music Pro", 9.99m);
        music.OnSubscriptionExpired += notifications.SendExpirationNotification;

        // Демонстрация новых методов
        streaming.ChangeQuality(2160);
        streaming.ToggleOfflineMode();
        streaming.AddFavoriteGenre("Фантастика");
        streaming.AddFeature("4K потокование");
        streaming.AddWatchTime(TimeSpan.FromHours(3.5));

        music.ToggleAudioQuality();
        music.CreatePlaylist("Мой топ");
        music.DownloadAlbum("Лучшие хиты 2024");
        music.ChangeListeningMode("Эквалайзер: Басы");
        music.AddFeature("Безлимитные загрузки");

        // Добавление в менеджер
        manager.AddSubscription(streaming);
        manager.AddSubscription(music);

        // Демонстрация изменения цены
        Console.WriteLine("\n--- Изменение цены ---");
        streaming.Cost = 12.99m;

        // Проверка коллекций
        Console.WriteLine("\n--- Активные подписки ---");
        foreach (var sub in manager.GetActiveSubscriptions())
        {
            Console.WriteLine($"Активная: {sub.ServiceName}");
        }

        Console.WriteLine($"\nОбщая ежемесячная стоимость: {manager.GetTotalMonthlyCost():C}");

        // Демонстрация работы с истечением срока
        Console.WriteLine("\n--- Тест истечения срока ---");
        var testSub = new Subscription("TEST", "Тестовая", 4.99m);
        testSub.OnSubscriptionExpired += notifications.SendExpirationNotification;
        testSub.ExpiryDate = DateTime.Now.AddSeconds(2); // Истечет через 2 секунды
        
        manager.AddSubscription(testSub);
        
        System.Threading.Thread.Sleep(3000); // Ждем 3 секунды
        manager.CheckAllExpirations();

        // Удаление подписки
        Console.WriteLine("\n--- Удаление подписки ---");
        manager.RemoveSubscription("TEST");

        // Показать все
        manager.DisplayAll();

        Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ЗАВЕРШЕНА ===");
        Console.WriteLine("Нажмите любую клавишу для выхода...");
        Console.ReadKey();
    }
}
```
